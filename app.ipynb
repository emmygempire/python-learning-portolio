QUESTION 3 
Create a function called countries_of_choice

The function takes input from the user

If the user input a name that matches the countries_of_choice, print it out.

If it does not, print Country is not available.

```python
favourite_countries = ["germany", "rusia", "Nigeria","Ghana", "Canada", "United Kingdom", "South Africa", "Kenya", "United States", "Germany"]


countries_of_choice = input()

print(countries_of_choice)


```

```python
if countries_of_choice in favourite_countries:

    print(f"{countries_of_choice}.")
else:
    print(f"{countries_of_choice} is not available.")
```
